<h2 style="color: magenta; font-family: Arial; font-style: italic;">
  Pydantic Basics: Creating and Using Models
</h2>


* Pydantic models are foundation of data validation in python.
* They use Python type annotations to define the structure and validate data runtime.

```python
# WITHOUT USING PYDANTIC:

from dataclasses import dataclass
```

```python
@dataclass
class Person():
    name : str
    age  : int
    city : str
```

```python
# creating instance/object

person_01 = Person(name='Sunita Mehta',age=20,city='surat')
print(person_01)
```

```python
person_02 = Person(name='Sunita Mehta',age=20,city=1232)
print(person_02)
```

<div style="font-family:Georgia; font-size:16px; border-left:5px solid purple; padding:15px; background-color:#e6f2ff; color:darkblue;">

<b>Explanation:</b><br>

When we use only the @dataclass decorator without inheriting from BaseModel, it doesn't enforce type checks.  
So even if city is defined as a str, passing a number (like in person_02) is still accepted — this can lead to unexpected bugs.

<b>Important Note:</b><br>

• @dataclass does not enforce type checking at runtime.  
• If we define city: str and pass 123, no error will be raised unless we manually validate or use a library like Pydantic.


<b>Pydantic Behavior:</b><br>

Using Pydantic's BaseModel enables automatic type checking and conversion at runtime.  
If we pass a wrong type, it either raises an error or converts it (e.g., number to string).


<b>Summary:</b><br>

• @dataclass = No type enforcement 

• BaseModel = Type-safe & auto-validating 

</div>


```python
# WITH PYDANTIC:

from pydantic import BaseModel   # BaseModel is a base class for creating Pydantic Models.
```

```python
# The class Person is inheriting the BaseModel.

class Person(BaseModel):
    
    name : str
    age  : int
    city : str
```

```python
# creating instance/object

person_01 = Person(name='Rashmi Mehta',age=20,city='surat')
print(person_01)
```

```python
type(person_01)
```

```python
person_02 = Person(name='Rashmi Mehta',age=20,city=390014)    # We Are Getting the ValidationError 
print(person_02)


# ValidationError:  1 validation error for Person , city Input should be a valid string [type=string_type, input_value=390014, input_type=int]
# It means: The field `city` expected a string, but got an integer (`390014`), so Pydantic raised a validation error.
```

<h3 style="font-style: italic; color: lime;">
 Note: Data validation occurred here because Pydantic enforces type checking and rejected the invalid input.
</h3>

<h2 style="color:magenta; font-family: Arial; font-style: italic;">
  Using Pydantic for Reliable Data Validation in Large Language Models (LLMs)
</h2>



**Challenges without Pydantic in LLM data validation:**

* Incorrect data types are accepted (e.g., numbers instead of strings).
* Missing or extra fields cause unexpected errors.
* Data inconsistencies are hard to detect early.
* Bugs and failures occur downstream due to bad input.
* Difficult to enforce consistent data formats.

---

**How LLMs benefit from using Pydantic:**

LLMs generate or process structured data like JSON or API inputs, which may be incorrect or incomplete. Pydantic automatically validates and enforces correct data types and formats at runtime, catching errors early and preventing invalid data from causing issues.

---

**Example:**
If an LLM outputs a user profile with age as the string "twenty-five" instead of an integer, Pydantic’s BaseModel will raise a validation error, stopping bad data from entering the system.

---

**Best Use Case:**
Validating and parsing LLM-generated data—especially for APIs, forms, or configurations—ensuring data integrity and reducing bugs by catching errors before further processing.



<h2 style="color: orange; font-family: Arial; font-style: italic;">
Pydantic: Required vs Optional Fields
</h2>

```python
from typing import Optional
```

```python
class Employee(BaseModel):
    
    id : int
    name : str
    department : str
    salary : Optional[float] = None   # Optional indicates that the fiel can be None ( Optional with Default Value)
    is_active : Optional[bool] = True # Optional with Default Value True 
```

```python
emp_01 = Employee(id=101,name='Abhi',department="IT")
print(emp_01)
```

```python
emp_02 = Employee(id=101,name='Abhi',department="IT",is_active=False)
print(emp_02)
```

```python
emp_03 = Employee(id=101,name='Abhi',department="IT",is_active=False,salary=600000)
print(emp_03)   # Here Automatic typecasting took place.
```

<div style="font-family: Georgia; font-size: 16px; border-left: 5px solid purple; padding: 15px; background-color: #e6f2ff; color: darkblue;">

<b>Definition :</b><br><br>

<ul>
<li><b>Optional[type]</b> : Indicates the field can be None</li>
<li><b>Default value (= None or = True)</b> : Makes the field Optional</li>
<li><b>Required fields</b> must still be provided</li>
<li><b>Pydantic validates types</b> even for optional fields when values are provided</li>
</ul>

<h2 style="color: orange; font-family: Arial; font-style: italic;">Pydantic and Collections: List, Tuple, Set, and Dict
 
</h2>

```python
# One of the field could be List.

from typing import List

class Classroom(BaseModel):
    room_number : str
    students    : List[str]  # List of strings
    capacity    : int
```

```python
# creating a classroom

classroom_with_List = Classroom(
    room_number="A101",
    students= ['Alice','Rash','Abhi','Ani','Anu','SK'],
    capacity=30
)

print(classroom_with_List)
```

```python
classroom_with_Tuple= Classroom(
    room_number="A101",
    students= ('Alice','Rash','Abhi','Ani','Anu','SK'),
    capacity=30
)

print(classroom_with_Tuple)   # Here we can see Typecasting is also Happening.
```

```python
classroom_with_Set= Classroom(
    room_number="A101",
    students= {'Alice','Rash','SK','Abhi','Ani','Anu','SK','SK','Anu','Abhi','aBHI','Anu'},
    capacity=20
)

print(classroom_with_Set)   # Here we can see Typecasting is also Happening.... # Since used set ==> No Dupicates.
```

```python
classroom_with_dict = Classroom(
    room_number="A101",
    students= {'name':'Hello','country' : 'India'},
    capacity=20
)

print(classroom_with_dict)  
```

<div style="font-family: Georgia; font-size: 16px; border-left: 5px solid purple; padding: 15px; background-color: #e6f2ff; color: darkblue;">

Pydantic expects a list of strings.

<b>What Worked and Why:</b>

* List – Matches exactly. ['Alice', 'Rash'] → Valid input.
* Tuple – Works due to type coercion.  ('Alice', 'Rash') → Pydantic auto-converts it to a list.
* Set – Also works due to auto-conversion, but: {'SK', 'SK'} → Converted to list, duplicates removed (because sets don’t allow them).

<b>What Didn't Work:</b>
* Dict – Invalid for List[str].

{'name': 'Hello', 'country': 'India'} is not iterable as a list of strings. → Throws validation error.

<b>Pydantic automatically casts compatible types (like tuple or set) to the expected list if possible.But dict does not match the structure of a list, so casting fails.</b>

```python
classroom_with_dict = Classroom(
    room_number="A101",
    students= [ {'name':'Hello','country' : 'India'} ],
    capacity=20
)

print(classroom_with_dict)  
```

Note : Because `students` is expected to be a `List[str]`, but you're passing a `List[Dict]` — a list of dictionaries instead of a list of strings.


```python
# What if Instead of List of String i provide  integer in List.... We will handle it using the try except block.

try:
    invalid_val = Classroom(room_number="A101",students=['SK',123],capacity=15)
except ValueError as e:
    print(e)
    
    
```

<h2 style="color: orange; font-family: Arial; font-style: italic;">
  Nested Models in Pydantic
</h2>

```python
class Address(BaseModel):
    street   : str
    city     : str
    zip_code : str
    

class Customer(BaseModel):
    customer_id : int
    name        : str
    address     : Address   # Nested Model
```

```python
# Create a Customer with Nested Address.
customer = Customer(
    customer_id=1,
    name = 'Abhi',
    address= {'street':"123 Main st" 
              , "city" : "Surat"
              , "zip_code": '390012'
            }
)
```

```python
customer
```

```python
# Question : What if i provide the 390012 as int instead of string.. Will this be typecasted automatically by pydantic? 
# Ans : No
```

```python
customer = Customer(
    customer_id=1,
    name = 'Abhi',
    address= {'street':"123 Main st" 
              , "city" : "Surat"
              , "zip_code": 390012
            }
)

print(customer)
```

```python
try:
    customer = Customer(
    customer_id=1,
    name = 'Abhi',
    address= {'street':"123 Main st" 
              , "city" : "Surat"
              , "zip_code": 390012
            })
    print(customer)
    
except ValueError as e:
    print(e)
    
```

```python
# But What if zip_code was in "int" and we provide in "str"..then  Yes... Automatic Typecasting.

class Address(BaseModel):
    street   : str
    city     : str
    zip_code : int
    

class Customer(BaseModel):
    customer_id : int
    name        : str
    address     : Address   # Nested Model
```

```python
customer = Customer(
    customer_id=1,
    name = 'Abhi',
    address= {'street':"123 Main st" 
              , "city" : "Surat"
              , "zip_code": 390012
            }
)

print(customer)
```

```python
customer = Customer(
    customer_id=1,
    name = 'Abhi',
    address= {'street':"123 Main st" 
              , "city" : "Surat"
              , "zip_code": "390012"
            }
)

print(customer)   # Here "390012" which was in str automatically got typecasted to int by pydantic.
```

<h2 style="color: orange; font-family: Arial; font-style: italic;">
 Field Customization and Validation in Pydantic
</h2>


<div style="font-family: Georgia; font-size: 16px; border-left: 5px solid purple; padding: 15px; background-color: #e6f2ff; color: darkblue;">

The Field() function in Pydantic is used to add extra rules or details to a model field, like setting default values, minimum/maximum limits, descriptions, or making a field optional.

The Field() function in Pydantic lets us customize fields by setting default values, constraints (like min/max), and adding descriptions.
If we want to enforce rules (e.g. age must be between 18–60), use Field() with arguments like ge, le, min_length, max_length, etc.


In Pydantic’s Field(), these keyword arguments are used to set numeric constraints:

- ge = greater than or equal to (e.g., ge=0 means value must be ≥ 0)
- le = less than or equal to
- gt = greater than
- lt = less than

```python
from pydantic import BaseModel , Field

class Item(BaseModel):
    name : str = Field(min_length=2,max_length=50)   # These are the constraints using Field() FUnction.
    price : float = Field(gt=0,le=1000) # gt means greater than 0 and le means less than or equal to 1000.
    quantity : int = Field(ge=0,description='Quantity Must be Non Negative.')        # ge means greater than or equal to.
    
```

```python
# Valid Instance:

item = Item(name='Book',price=29.99,quantity=10)
print(item)
```

```python
# InValid Instance:

try: 
    item = Item(name="Book",price=1200,quantity=-1)
    print(item)
    
except ValueError as e:
    print(e)
```

```python
class User(BaseModel):
    username : str = Field(...,description="Unique username for the user")
    age      : int = Field(default=18,description="User age , defaults to 18")
    email    : str = Field(default_factory= lambda : "user@example.com" , description="Default Email address")
    
```

```python
user1 = User(username="Rash")
print(user1)
```

```python
user2=User(username="Bob",age=25,email='bob@gmail.com')
print(user2)
```
